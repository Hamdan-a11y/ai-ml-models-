```python
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
import joblib
from  sklearn.metrics import accuracy_score, r2_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
d = pd.read_csv("/kaggle/input/student-habits-vs-academic-performance/student_habits_performance.csv")
LE = LabelEncoder()
d['gender'] = LE.fit_transform(d['gender'])
d['part_time_job'] = LE.fit_transform(d['part_time_job'])
d['diet_quality'] = LE.fit_transform(d['diet_quality'])
d['parental_education_level'] = LE.fit_transform(d['parental_education_level'])
d['internet_quality'] = LE.fit_transform(d['internet_quality'])
d['extracurricular_participation'] = LE.fit_transform(d['extracurricular_participation'])
d=d.drop('student_id', axis=1)
x=d.drop('mental_health_rating', axis=1)

y=d['mental_health_rating']

s=StandardScaler()
x=s.fit_transform(x)

x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2, random_state=42)

model=GradientBoostingClassifier(n_estimators=100)
model.fit(x_train, y_train)
predicted_values = model.predict(x_test, )
y_pred = model.predict(x_test)
#print(r2_score(y_test,y_pred))
print(accuracy_score(y_test, predicted_values))
print(classification_report(y_test, predicted_values))

joblib.dump(model, 'model.pkl')
joblib.dump(s, 'scaler.pkl')
joblib.dump(LE, 'encoder.pkl')
```
