```
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import joblib
import matplotlib.pyplot as plt

# Load Dataset
d=pd.read_csv("/kaggle/input/weather-dataset/weatherHistory.csv")
d=d.drop('Formatted Date',axis=1)

c_s=LabelEncoder()
d['Summary']=c_s.fit_transform(d['Summary'])
d['Precip Type']=c_s.fit_transform(d['Precip Type'])
d['Daily Summary']=c_s.fit_transform(d['Daily Summary'])
d = d.sample(5000, random_state=42)


x=d.drop('Temperature (C)',axis=1)
y=d['Temperature (C)']


scaler=StandardScaler()
x=scaler.fit_transform(x)
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)


model_SVM=SVR()


model_SVM.fit(x_train,y_train)

svm_y=model_SVM.predict(x_test)
print("MAE:", mean_absolute_error(y_test, svm_y))
print("MSE:", mean_squared_error(y_test, svm_y))
print("R2 Score:", r2_score(y_test, svm_y))

import joblib
joblib.dump(model_SVM,"model.pkl")
joblib.dump(scaler,"scaler.pkl")
```
